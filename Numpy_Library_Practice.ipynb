```python
#Task 1:Student Performance Analyzer
import numpy as np

marks = np.array([
    [78, 85, 90, 66],
    [55, 72, 68, 80],
    [92, 88, 95, 90],
    [45, 60, 52, 48],
    [70, 75, 65, 82]
])

student_total = np.sum(marks,1)
print("Student Total Marks:", student_total)

student_average = np.mean(marks,1)
print("Students Average Marks:", student_average)

highest_total = np.max(student_total)
print("Student Highest Total Marks:",highest_total)

lowest_total = np.min(student_total)
print("Studnet Lowest Total Marks:",lowest_total)

passed = marks[marks>= 50]
print("Passed Subject Marks:",passed)

fail= marks[marks<50]
print("Fail Subject Marks:",fail)

above_70 = marks[marks>=70]
print("Subject Above 70 Marks:",above_70)

highest_average = np.max(student_average)
print("Highest Average Marks:",highest_average)

sub_average = np.mean(marks,0)
print("Subject Average Marks:",sub_average)

highest_subaverage = np.max(sub_average)
print("Highest Subject Average Marks:",highest_subaverage)

increase_marks = marks +(marks*5/100)
print("After 5% Increased Marks:",increase_marks)

rounded_marks = np.round(increase_marks,2)
print("Round off marks:",rounded_marks)

excellent_marks = marks[marks>=80]
print("Excellent Marks:",excellent_marks)

need_improvement = marks[marks<80]
print("Need Improvement:",need_improvement)
```
