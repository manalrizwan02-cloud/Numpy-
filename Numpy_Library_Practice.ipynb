```python
# Task 1: Student Marks Analyzer
import numpy as np

Stds_marks = np.array([50,70,80,90,10])
print("Average marks:", np.mean(Stds_marks))
print("Highest marks:",np.max(Stds_marks))
print("Lowest Marks:",np.min(Stds_marks))
print("Total Passed Students :",np.sum(Stds_marks >= 50))
print("Total failed Students:",np.sum(Stds_marks < 50))
```
